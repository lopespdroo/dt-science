```python
"""Showing how function works in python"""
def double(x):
    return x* 2
```

```python
"""First-class functions in actions.
    With them I can call a function 
    without executing it right-away."""
def apply_to_one(f):
    return f(1)

my_double = double
x = apply_to_one(my_double)

x
```

```python
"""Lambda functions"""
y = apply_to_one(lambda x: x + 4)
y
```

```python
"""Counters"""

document = ["janeiro",
            "janeiro",
            "janeiro",
            "janeiro",
            "fevereiro",
            "fevereiro",
            "fevereiro",
            "fevereiro",
            "março",
            "abril",
            "maio",
            "junho",
            "junho",
            "junho",
            "junho",
            "julho"
            "agosto"
            "setembro",
            "setembro",
            "outubro",
            "novembro",
            "dezembro",
            "dezembro",
            "dezembro"
           ]

print(document)
```

```python
from collections import defaultdict

word_counts = defaultdict(int)
print(word_counts)
```

```python
"""Implemeting a counter with defaultdict"""
for word in document:
    word_counts[word] += 1

word_counts
```

```python
"""Using the Counter"""
from collections import Counter

word_counts_with_counter = Counter(document)
print(word_counts_with_counter)
```

```python
"""Sets"""

document_set = set(document)
document_set
```

```python
"""Sorting functions and use cases"""
x = [4, 1, 2, 3]
y = sorted(x)

print(y)
x.sort()
print(x)

x = sorted([-4, 1, -2, 3], key=abs, reverse=True) #reverse with absolute value
z = sorted([-4, 1, -2, 3], reverse=True) #"normal" reverse

print(x)
print(z)
```

```python
"""List compressions"""

even_numbers = [x for x in range(5) if x % 2 == 0]
squares = [x * x for x in range(5)]
even_squares = [x * x for x in even_numbers]

print(even_numbers)
print(squares)
print(even_squares)
```

```python
"""Making list compressions using lambda functions"""
new_even_numbers = [x for x in range(10) if x % 2 == 0]
square = (lambda x: x*x)
new_even_squares = [square(x) for x in new_even_numbers]

new_even_squares
```

```python
"""Transforming lists in dicts and sets"""
square_dict = {x: x * x for x in range(5)}
square_set = {x * x for x in [1, -1]}

print(square_dict)
print(square_set)
```

```python
"""Asserts"""

def smallest_item(xs):
    return min(xs)

assert smallest_item([10, 20, 5, 40]) == 5, "smallest item should be 5"
assert smallest_item([1, 0, -1, 2]) == -1

smallest_item([10, 20, 5, 40])
```
